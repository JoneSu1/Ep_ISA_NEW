# Ep_ISA_NEW all reconstructed figures

This notebook regenerates the full figure set from locally extracted Google Drive data. It only loads precomputed tables and Fi-NeMo hit files; it does not run model inference, motif ablation, null sampling, or any new ISA computation.

Input snapshot: `F:/phd/Drophila/3Model_motif_discovering/ISA/result/google_drive_snapshot`  
Output directory: `F:/phd/Drophila/3Model_motif_discovering/ISA/result/all_reconstructed_figures_from_gdrive`

## Figure set

Main figures:

1. Fig. 1: sequence-level CAGE-with-E/D classification and motif position architecture.
2. Fig. 2: real-vs-null single ISA and pair interaction.
3. Fig. 3: cooperativity ECDF and absolute sign composition.
4. Fig. 4: distance context control, isolated vs embedded motif pairs.
5. Fig. 5: CAGE-with-E/D recurrent pair reuse and direction switching.
6. Fig. 6: TF-level grammar, importance and partner specificity.

Extended figures:

- Extended 1: full per-task TF-pair heatmaps.
- Extended 2: n_effective, median distance, and q-value QC.
- Extended 3: old relative category label vs absolute sign audit.
- Extended 4: motif position distributions and TF-level median position heatmap.

```python
from pathlib import Path
import sys

REPO_DIR = Path(r"F:/phd/Drophila/3Model_motif_discovering/ISA/Ep_ISA_NEW")
SNAPSHOT_DIR = Path(r"F:/phd/Drophila/3Model_motif_discovering/ISA/result/google_drive_snapshot")
OUT_DIR = SNAPSHOT_DIR.parent / "all_reconstructed_figures_from_gdrive"
MAIN_DIR = OUT_DIR / "main_figures"
EXT_DIR = OUT_DIR / "extended_figures"
for d in [OUT_DIR, MAIN_DIR, EXT_DIR]:
    d.mkdir(parents=True, exist_ok=True)

SRC_DIR = REPO_DIR / "src"
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from Ep_ISA_NEW.plotting.reconstructed import (
    setup_nature_style,
    load_precomputed_results,
    load_finemo_hits,
    summarize_loaded_results,
    recurrent_pair_sets,
    plot_sequence_cage_with_ed_overview,
    plot_real_vs_null_composite,
    plot_coop_ecdf_and_sign,
    plot_distance_context,
    plot_recurrent_heatmap_and_switch,
    plot_recurrent_paired_differences,
    plot_recurrent_combined_figure,
    plot_tf_level_summary,
    plot_extended_full_heatmaps,
    plot_extended_quality_metrics,
    plot_extended_category_audit,
    plot_extended_motif_position_by_tf,
)

arial_path = setup_nature_style()
print(f"Arial font: {arial_path}")
print(f"Snapshot: {SNAPSHOT_DIR}")
print(f"Output: {OUT_DIR}")
```

## 1. Load precomputed data

```python
results = load_precomputed_results(SNAPSHOT_DIR)
hits = load_finemo_hits(SNAPSHOT_DIR)
summary = summarize_loaded_results(results)
summary.to_csv(OUT_DIR / "loaded_google_drive_snapshot_summary.csv", index=False)
summary
```

```python
pair_set_sizes = {key: len(value) for key, value in recurrent_pair_sets(results).items()}
pair_set_sizes
```

```python
tf_summary, partner_specificity = plot_tf_level_summary(
    results,
    outpath=MAIN_DIR / "fig6_tf_level_grammar.pdf",
)
tf_summary.to_csv(MAIN_DIR / "fig6_tf_importance_coop_summary.csv", index=False)
partner_specificity.to_csv(MAIN_DIR / "fig6_partner_specificity.csv", index=False)
tf_summary.head()
```
